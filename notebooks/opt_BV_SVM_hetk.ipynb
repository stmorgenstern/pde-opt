```python
binary_mask = np.zeros((200, 200))
center = (100, 100)
radius = 75
y, x = np.ogrid[:200, :200]
dist_from_center = np.sqrt((x - center[0]) ** 2 + (y - center[1]) ** 2)
binary_mask[dist_from_center <= radius] = 1

shape = Shape(
    binary=jnp.array(binary_mask),
    dx=(1.0, 1.0),
    smooth_epsilon=3.0,
    smooth_curvature=0.008,
    smooth_dt=0.01,
    smooth_tf=100.0,
)
```

```python
Nx, Ny = binary_mask.shape
pixel_length_rat = 50e-9
Lx=Nx*pixel_length_rat
Ly = Ny*pixel_length_rat
# Lx = 0.01 * Nx
# Ly = 0.01 * Ny

domain = Domain(
    (Nx, Ny), ((-Lx / 2, Lx / 2), (-Ly / 2, Ly / 2)), "dimensionless", shape
)
```

```python
kappa = 0.002 #dimensionless
c_max = 2.29e4 #mol/m^3
F = 9.64853321233100184e4 #C⋅mol−1.
k0 = (1e-2 / F) #mol/(m^2 s)
H = 150e-9 #m
C_rate = 1e-2; #1/hr
kappa = 3.1314026858813996e13 #eV/m
```

```python
kB = 8.617333262145e-5; #Boltzmann Constant [eV/K]
T=298.15; #Temperature [K]
omega = 0.11484582867125136; #eV
omega_tilde = omega / (kB * T)
f=lambda c: c * jnp.log(c) + (1.0 - c) * jnp.log(1.0 - c) + omega_tilde * c * (1.0 - c) + 0.059
mu=lambda c: jnp.log(c / (1.0 - c)) + omega_tilde * (1.0 - 2.0 * c)
j0 = lambda c: jnp.sqrt(c)*(1.0 - c)
```

```python
# ----------------------------
# 0) Build mean-constrained evaluator and PDE instance
# ----------------------------
Nb = 100
sh = SpatialHeterogeneity.from_shape(shape=domain.geometry, n_basis=Nb, prior_fn=jnp.exp)
sh.build_mean_constrained(mask_type="binary")
k_eval_mc = sh.make_mean_constrained_evaluator(apply_mask=False)

theta0 = jnp.zeros((sh.mean_constrained_U.shape[1],))

eq = AllenCahn2DSBM_BV_CC(
    domain=domain, kappa=kappa, f=f, mu=mu, j0=j0,
    k0=k0, H=H, Cmax=c_max, Crate=C_rate,
    use_prefactor=True, mean_constrained=True,
    k_eval=None, k_eval_mc=k_eval_mc, theta=theta0,
)

# ----------------------------
# 1) Initial condition (your nucleation map)
# ----------------------------
shape = domain.geometry
psi_edge_inner = shape.get_mask_edge(connectivity=8, side="inner")
nucsite = jnp.where(psi_edge_inner, 0.2, 0.0)
key = random.PRNGKey(0)
u0_edge = gen_c0(shape, key, noise_fac=0.0, nucsite=nucsite)

# ----------------------------
# 2) Warmup solve with YOUR event to bound tf
# ----------------------------
t_start = 0.0
t_final_guess = eq.tf_tilde_est
dt0 =(eq.tf_tilde_est - t_start) / 100.0
def make_soc_event(eqmodel, *, threshold: float = 0.999):
    def cond_fn(t, y, args, **kwargs):
        return eqmodel.get_SOC(y) > threshold
    return dfx.Event(cond_fn)

event = make_soc_event(eq, threshold=0.999)
solver = ROCK2JAX()

warm_term = dfx.ODETerm(lambda t, y, args: eq.rhs(y, t))
warm_sol = dfx.diffeqsolve(
    warm_term, solver,
    t0=t_start, t1=t_final_guess, dt0=dt0, y0=u0_edge,
    stepsize_controller=dfx.PIDController(rtol=1e-4, atol=1e-6)
    , max_steps=50_000, event=event,
)
t_final_bound = warm_sol.ts[-1]
```

```python
# Fixed time grid for optimization runs (no event)
Nsave = 128
ts_fixed = jnp.linspace(t_start, t_final_bound, Nsave)

# ----------------------------
# 3) Forward solve on fixed horizon (no event), pass k_field via args
# ----------------------------
rhs_with_k = eq.rhs_with_k_fn           # jitted function: (y, t, k_field) -> dy/dt
fixed_ode   = dfx.ODETerm(lambda t, y, k_field: rhs_with_k(y, t, k_field))
controller  = dfx.PIDController(rtol=1e-4, atol=1e-6)

def forward_fixed(y0, ts, k_field):
    dt0 = (ts[-1] - ts[0]) / 100
    return dfx.diffeqsolve(
        fixed_ode, solver,
        t0=ts[0], t1=ts[-1], dt0=dt0, y0=y0,
        args=k_field,
        stepsize_controller=controller,
        saveat=dfx.SaveAt(ts=ts),
        max_steps=50_000, event=None,
    )
forward_fixed_jit = eqx.filter_jit(forward_fixed)

# ----------------------------
# 4) Variance objective (time-averaged) — arrays only
# ----------------------------
binmask = domain.geometry.binary

def soc_local(u, m):     # mean over mask
    return jnp.sum(u * m) / jnp.sum(m)

def cvar_local(u, m):    # variance over mask
    cbar = soc_local(u, m)
    return jnp.sum((u - cbar)**2 * m) / jnp.sum(m)

def objective_from_solution(ys, m):
    # ys: (T, Nx, Ny)
    cvars = jax.vmap(lambda u: cvar_local(u, m))(ys)
    return jnp.sum(cvars)

# Optimistix objective signature: (theta, args) -> scalar
def bfgs_objective(theta, args):
    y0, ts, m = args
    k_field = k_eval_mc(theta)                      # mean-constrained k(x,y)
    sol = forward_fixed_jit(y0, ts, k_field)        # fixed horizon
    return objective_from_solution(sol.ys, m)

```

```python
# -------- UC pieces (psi-weighted) --------
def _uc_from_ys(ys, psi):
    """
    ys:  (T, Nx, Ny)
    psi: (Nx, Ny) weighting field (domain.geometry.smooth or .binary)
    """
    w = jnp.asarray(psi)
    A = jnp.sum(w)  # dxdy cancels, so we can omit it

    # mean concentration per time
    cbar_t = jnp.sum(ys * w[None, ...], axis=(1, 2)) / A  # (T,)

    # variance per time
    diffs = ys - cbar_t[:, None, None]
    v_t = jnp.sum((diffs**2) * w[None, ...], axis=(1, 2)) / A  # (T,)

    # UC = 1 - sum_i sqrt(v_i * cbar_i (1-cbar_i)) / sum_i cbar_i (1-cbar_i)
    eps = 1e-12
    cterm = jnp.clip(cbar_t * (1.0 - cbar_t), 0.0, 1.0)  # safety
    numer = jnp.sum(jnp.sqrt(jnp.maximum(v_t, 0.0) * cterm))
    denom = jnp.sum(cterm) + eps
    UC = 1.0 - numer / denom
    return UC

# -------- Optimistix objective: minimize (1 - UC) --------
# args will be (y0, ts_fixed, psi)
def uc_objective(theta, args):
    y0, ts, psi = args
    k_field = k_eval_mc(theta)                 # mean-constrained k(x,y)
    sol = forward_fixed(y0, ts, k_field)       # fixed-horizon solve (no event)
    UC = _uc_from_ys(sol.ys, psi)
    return 1.0 - UC                            # minimize 1-UC  <=> maximize UC


# Choose psi for quadrature: smooth (recommended) or binary
psi = domain.geometry.smooth  # or: domain.geometry.binary
```

```python


# ---- Run BFGS on theta ----
solver_bfgs = optx.BFGS(
    rtol=1e-4, atol=1e-6,
    verbose=frozenset({"step","accepted","loss","step_size"}),
)
res_uc = optx.minimise(
    uc_objective,
    solver_bfgs,
    theta0,
    args=(u0_edge, ts_fixed, psi),   # arrays-only args (no eq/domain objects)
    max_steps=200,
    throw=False,
)

theta_star_uc = res_uc.value
print("BFGS status:", res_uc.state)

# ---------- Report UC (init vs optimized) ----------
# compute baseline and optimized UC on the SAME fixed horizon
sol_init = forward_fixed(u0_edge, ts_fixed, k_eval_mc(theta0))
sol_star = forward_fixed(u0_edge, ts_fixed, k_eval_mc(theta_star_uc))

UC_init = float(_uc_from_ys(sol_init.ys, psi))
UC_star = float(_uc_from_ys(sol_star.ys, psi))
print(f"UC initial:   {UC_init:.6f}")
print(f"UC optimized: {UC_star:.6f}")
```

```python
# --- EMA-wrapped Armijo that plugs into your BFGS ---

import dataclasses
from typing import Any, Tuple
import equinox as eqx
import jax.numpy as jnp
import optimistix as optx
from jaxtyping import Array, Bool

# Carry both the inner state and the EMA value
@dataclasses.dataclass
class _EMAState(eqx.Module):
    inner_state: Any
    mu: float

class EMABacktrackingArmijo(
    optx.AbstractSearch[
        Any,  # Y
        optx.FunctionInfo,             # _FnInfo (EvalGrad/Hess/HessInv etc.)
        optx.FunctionInfo.Eval,        # _FnEvalInfo (Eval)
        _EMAState                      # State
    ],
    strict=True,
):
    inner: optx.BacktrackingArmijo
    beta: float = 0.9           # EMA decay
    mu_init: float = 1.0        # first-iter guess
    mu_min: float = 1e-8
    mu_max: float = 10.0
    bold: float = 1.0           # >1.0 to probe a bit larger than EMA

    # Match your search.init(y, f_info_struct) signature
    def init(self, y, f_info_struct) -> _EMAState:
        inner_state = self.inner.init(y, f_info_struct)
        return _EMAState(inner_state=inner_state, mu=self.mu_init)

    # Match your search.step(...) -> (step_size, accept, RESULTS, state)
    def step(
        self,
        first_step: Bool[Array, ""],
        y,
        y_eval,
        f_info,
        f_eval_info: optx.FunctionInfo.Eval,
        state: _EMAState,
    ) -> tuple[float, Bool[Array, ""], optx.RESULTS, _EMAState]:
        # Use EMA as this iteration's step_init (optionally "bolded")
        mu = jnp.clip(self.bold * state.mu, self.mu_min, self.mu_max)
        inner_now = eqx.tree_at(lambda s: s.step_init, self.inner, mu)

        alpha, accept, result, new_inner_state = inner_now.step(
            first_step, y, y_eval, f_info, f_eval_info, state.inner_state
        )

        # Update EMA only when a step is actually accepted
        new_mu = jnp.where(accept, self.beta * state.mu + (1.0 - self.beta) * alpha, state.mu)
        new_state = _EMAState(inner_state=new_inner_state, mu=float(new_mu))
        return float(alpha), accept, result, new_state

# 1) Build the base Armijo with your API (slope/decrease_factor/step_init)
armijo = optx.BacktrackingArmijo(
    decrease_factor=0.5,   # backtracking shrink
    slope=1e-4,            # Armijo c1
    step_init=1.0
)

# 2) Wrap with EMA logic
ema_search = EMABacktrackingArmijo(
    inner=armijo,
    beta=0.9,
    mu_init=1.0,
    mu_min=1e-8,
    mu_max=10.0,
    bold=1.0,              # e.g. 1.25 if you want to probe a bit larger
)

# 3) Make your solver and PATCH its .search field
solvert = optx.BFGS(rtol=1e-4, atol=1e-6, verbose=frozenset({"step","accepted","loss","step_size"}))
solvert = eqx.tree_at(lambda s: s.search, solvert, ema_search)


```

```python
import jax.numpy as jnp

def uniformity_coefficient(solution, psi):
    """
    solution: diffrax solution with fields .ys (T, Nx, Ny) and .ts (T,)
    psi: area-weighting field (use domain.geometry.smooth or .binary)
    """
    ys = solution.ys                      # (T, Nx, Ny)
    w  = jnp.asarray(psi)                 # (Nx, Ny)
    A  = jnp.sum(w)                       # ∫Ω dA  ~  sum(psi) * dxdy  (dxdy cancels)

    # mean concentration over Ω at each t
    cbar_t = jnp.sum(ys * w[None, ...], axis=(1, 2)) / A             # (T,)

    # variance over Ω at each t
    diff  = ys - cbar_t[:, None, None]
    v_t   = jnp.sum((diff**2) * w[None, ...], axis=(1, 2)) / A       # (T,)

    # UC = 1 - sum_i sqrt(v(t_i) * cbar(t_i) * (1 - cbar(t_i))) / sum_i cbar(t_i) * (1 - cbar(t_i))
    # add tiny eps for numerical safety if cbar ~ 0 or 1
    eps   = 1e-12
    cterm = jnp.clip(cbar_t * (1.0 - cbar_t), 0.0, 1.0)
    numer = jnp.sum(jnp.sqrt(jnp.maximum(v_t, 0.0) * cterm))
    denom = jnp.sum(cterm) + eps
    UC    = 1.0 - numer / denom
    return UC, cbar_t, v_t
psi = domain.geometry.smooth  # or domain.geometry.binary

UC_init, cbar_init, v_init = uniformity_coefficient(sol_init, psi)
UC_star, cbar_star, v_star = uniformity_coefficient(sol_star, psi)

print(f"UC (initial):  {float(UC_init):.6f}")
print(f"UC (optimized):{float(UC_star):.6f}")
```
